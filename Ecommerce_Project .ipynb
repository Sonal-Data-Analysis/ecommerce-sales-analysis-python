
# Ecommerce Sales Analysis Using Python

## Project Objective:
                   The objective of this project is to analyze sales and profit data using Python, Pandas, and Plotly.
                   The analysis focuses on monthly trends, product categories, sub-categories 
                   and customer segments to understand business performance and identify areas for improvement.

## Import Required Libraries.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors

pio.templates.default = "plotly_white"
```

## Create Year and Month Columns

```python
data['Order Month']=data['Order Date'].dt.month
data['order Year']=data['Order Date'].dt.year
data['week of day']=data['Order Date'].dt.dayofweek
```

```python
data
```

# Monthly Sales Analysis

```python

sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()

fig = px.line(sales_by_month, x='Order Month', y='Sales',
              title='Monthly Sales Analysis')

fig.show()

```

# Sales by category

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()

fig = px.pie(sales_by_category, names='Category', values='Sales',
             title='Sales by Category')

fig.show()
```

# Sales Analysis by Sub Category

```python

sales_by_category = data.groupby('Sub-Category')['Sales'].sum().reset_index()

fig = px.bar(sales_by_category, names='Sub-Category', values='Sales',
             title='Sales by Sub-Category')

fig.show()
```

```python

sales_by_category = data.groupby('Sub-Category')['Sales'].sum().reset_index()

fig = px.bar(sales_by_category, x='Sub-Category', y='Sales',
             title='Sales by Sub-Category')

fig.show()
```

# Monthly Profit Analysis 

```python

monthly_profit = data.groupby(['Order Month'])['Profit'].sum().reset_index()

fig = px.line(
    monthly_profit,
    x='Order Month',
    y='Profit',
    title='Monthly Profit Analysis',
    markers=True
)

fig.show()
```

# Profit by category

```python
profit_data = data.groupby('Category')['Profit'].sum().reset_index()

fig = px.pie(profit_data, names='Category', values='Profit',
             title='Profit by Category')

fig.show()
```

# Profit by Subcategory

```python
profit_data = data.groupby('Sub-Category')['Profit'].sum().reset_index()

fig = px.bar(profit_data, x='Sub-Category', y='Profit',
             title='Profit by Sub-Category')

fig.show()
```

# Sales and Profit - Customer Segment

```python

segment_data = data.groupby('Segment')[['Sales', 'Profit']].sum().reset_index()

fig = px.bar(
    segment_data,
    x='Segment',
    y=['Sales', 'Profit'],
    title='Sales and Profit by Customer Segment',
    barmode='group'
)

fig.show()
```

# Sales to Profit ratio

```python

sales_profit_by_segment = data.groupby('Segment')[['Sales', 'Profit']].sum().reset_index()

sales_profit_by_segment['Sales to Profit Ratio'] = (
    sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
)

print(sales_profit_by_segment[['Segment', 'Sales to Profit Ratio']])
```
